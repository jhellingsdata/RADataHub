```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles   # adjust import to match your package

styles = EcoStyles()
styles.register_and_enable_theme()

SOURCE = ("Source: OECD National Accounts, Table 12 (government deficit/surplus, "
          "revenue, expenditure and main aggregates), accessed January 2026")
EXPLAINER = ("Net lending (+) / net borrowing (−) is revenue minus spending, "
             "including investment, as a share of GDP.")

TEAL, GREY = "#00a19b", "#b8bcc8"   # swap for your ecostyles teal / grey

# ------------------------------------------------------------------
# RAW DATA (copied from Table 18, % of GDP, values as in the table)
# ------------------------------------------------------------------

# United Kingdom, 1990-2024
uk_years = list(range(1990, 2025))
uk_central = [-1.45, -2.81, -6.65, -7.53, -6.44, -5.04, -3.95, -1.72, 0.15, 1.02,
              1.23, 0.60, -1.87, -3.30, -2.90, -2.78, -2.66, -2.51, -4.78, -9.50,
              -9.01, -7.12, -7.48, -5.17, -5.57, -4.44, -2.89, -2.00, -1.89, -1.92,
              -13.07, -7.66, -4.17, -5.60, -5.53]
uk_local = [-0.16, -0.17, 0.69, -0.01, -0.24, -0.37, -0.15, -0.24, -0.36, -0.26,
            0.16, -0.26, -0.07, 0.11, -0.24, -0.35, -0.08, -0.12, -0.27, -0.47,
            -0.24, -0.30, -0.48, -0.12, 0.10, -0.09, -0.39, -0.48, -0.34, -0.46,
            0.16, 0.12, -0.43, -0.45, -0.53]
assert len(uk_years) == len(uk_central) == len(uk_local)

# ------------------------------------------------------------------
# Chart 2: UK central vs local government, 1990-2024
# ------------------------------------------------------------------
uk = pd.DataFrame({"year": uk_years,
                   "Central government": uk_central,
                   "Local government": uk_local})
c2 = uk.melt(id_vars="year", var_name="level", value_name="balance")
c2["date"] = pd.to_datetime(c2["year"].astype(str), format="%Y")

key = alt.Legend(
    title=None,
    orient="bottom",            # floating key, as in your canonical style
    direction="horizontal",
    legendX=0, legendY=-32,   # above the plot, left-aligned
    symbolType="stroke",      # line swatch, not a square
    symbolStrokeWidth=3,
    symbolSize=300,
    labelFontSize=12,
    columnPadding=16,
)

chart2 = (
    alt.Chart(c2)
    .mark_line(strokeWidth=1.8)
    .encode(
        x=alt.X("date:T", axis=alt.Axis(format="%Y", tickCount=8), title=None),
        y=alt.Y("balance:Q", title="% of GDP"),
        color=alt.Color("level:N", legend=key,
                        scale=alt.Scale(domain=["Central government", "Local government"],
                                        range=[GREY, TEAL])),
    )
    .properties(
        width=400, height=300,
    )
)

styles.save(chart2, path=".", name="uk_net_lending_by_level", svg=True)
```
