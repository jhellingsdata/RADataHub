```python
import altair as alt
import pandas as pd
from ecostyles import EcoStyles

styles = EcoStyles()
styles.register_and_enable_theme()

def d(year):
    return pd.Timestamp(f"{year}-01-01")

# ---- Data: bonds issued, then approved borrowing to the 2026 forecast peak, £m -------
actual = pd.DataFrame({
    "date": [d(2010), d(2014), d(2022)],
    "value": [0, 250, 750],           # 2014 housing bond; 2022 £500m pension bond
})
forecast = pd.DataFrame({
    "date": [d(2022), d(2026)],
    "value": [750, 1300],             # + new hospital (£523m) and Fort Regent (£43m)
})

# ---- Chart -----------------------------------------------------------------------
Y_TITLE = "Borrowing, \u00a3m"
y_scale = alt.Scale(domain=[0, 1500])

x = alt.X("date:T", title=None, scale=alt.Scale(domain=[d(2010), d(2029)]),
          axis=alt.Axis(format="%Y", tickCount=5))

def y(field):
    return alt.Y(f"{field}:Q", title=Y_TITLE, scale=y_scale, axis=alt.Axis(format=",.0f"))

colour = alt.Color("series:N", legend=None)  # theme assigns the palette

def tag(chart):
    return chart.transform_calculate(series="'Debt'")

area = tag(alt.Chart(actual)).mark_area(interpolate="step-after", opacity=0.15).encode(
    x=x, y=y("value"), color=colour
)
line = tag(alt.Chart(actual)).mark_line(interpolate="step-after", strokeWidth=1.8).encode(
    x=x, y=y("value"), color=colour
)
ramp_area = tag(alt.Chart(forecast)).mark_area(opacity=0.07).encode(
    x=x, y=y("value"), color=colour
)
ramp = tag(alt.Chart(forecast)).mark_line(strokeWidth=1.8, strokeDash=[5, 4]).encode(
    x=x, y=y("value"), color=colour
)

def label(rows, **mark):
    df = pd.DataFrame(rows)
    return tag(alt.Chart(df)).mark_text(**mark).encode(
        x=x, y=y("value"), text="text:N", color=colour
    )

# Amounts only: what the borrowing was for goes in the slide text
first = label([{"date": d(2014), "value": 250, "text": "\u00a3250m"}],
              align="left", dx=6, dy=-9, fontWeight="bold")
second = label([{"date": d(2022), "value": 750, "text": "\u00a3750m"}],
               align="right", dx=-8, fontWeight="bold")
peak = label([{"date": d(2026), "value": 1300, "text": "\u00a31.3bn"}],
             align="left", dx=8, fontSize=22, fontWeight="bold")

chart = (area + ramp_area + line + ramp + first + second + peak).properties(
    width=700, height=270,
    autosize=alt.AutoSizeParams(type="fit", contains="padding"),
)

styles.save(chart, path=".", name="jersey_debt", svg=True)
chart
```
