```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

df = pd.DataFrame({
    "decile": list(range(1, 11)),
    "share": [0, 0.03, 0, 0.34, 0.27, 1.64, 8.36, 4.51, 21.12, 63.72],
})

df["series"] = "x"

bars = alt.Chart(df).mark_bar(cornerRadius=2).encode(
    x=alt.X("decile:O", title="Household wealth decile (1 = least wealthy, 10 = most)",
            axis=alt.Axis(labelAngle=0)),
    y=alt.Y("share:Q", title="Share of directly-held UK bonds and gilts (%)",
            scale=alt.Scale(domain=[0, 70])),
    color=alt.Color("series:N", legend=None))


labels = alt.Chart(df[df.share >= 1]).mark_text(dy=-6, fontSize=11, color="#122b39").encode(
    x="decile:O", y="share:Q", text=alt.Text("share:Q", format=".0f"))

caption = alt.Title(
    text="Source: author's calculations, ONS Wealth and Assets Survey (Round 8)",
    subtitle=[
        "Share of all directly-held UK bonds and gilts, by household wealth decile, %.",
        "The wealthiest tenth of households hold nearly two-thirds.",
    ],
    orient="bottom", anchor="start", fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (bars + labels).properties(width=640, height=340, title=caption).configure(
    background="white", font="Circular Std").configure_view(
    fill="transparent", stroke="transparent")

styles.save(chart, path=".", name="gilt_by_wealth_decile", svg=True)
chart.save("gilt_by_wealth_decile.png", scale_factor=2.0)
chart
```
