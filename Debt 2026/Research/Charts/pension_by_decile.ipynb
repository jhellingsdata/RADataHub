```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

df = pd.DataFrame({
    "decile": list(range(1, 11)),
    "share": [0.12, 0.45, 1.73, 2.95, 4.94, 6.50, 8.88, 12.66, 19.96, 41.81],
})
df["hi"] = df["decile"] == 10

bars = alt.Chart(df).mark_bar(cornerRadius=2).encode(
    x=alt.X("decile:O", title="Household wealth decile (1 = least wealthy, 10 = most)",
            axis=alt.Axis(labelAngle=0)),
    y=alt.Y("share:Q", title="Share of total pension wealth (%)", scale=alt.Scale(domain=[0, 45])),
    color=alt.Color("hi:N", scale=alt.Scale(domain=[True, False], range=["#E6224B", "#36B7B4"]),
                    legend=None))
labels = alt.Chart(df[df.share >= 1]).mark_text(dy=-6, fontSize=11, color="#122b39").encode(
    x="decile:O", y="share:Q", text=alt.Text("share:Q", format=".0f"))

caption = alt.Title(
    text="Source: author's calculations, ONS Wealth and Assets Survey (Round 8)",
    subtitle=[
        "Share of total household pension wealth, by wealth decile, %.",
        "Pension wealth — the main indirect channel into gilts — is also concentrated at the top.",
    ],
    orient="bottom", anchor="start", fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (bars + labels).properties(width=640, height=340, title=caption).configure(
    background="white", font="Circular Std").configure_view(fill="transparent", stroke="transparent")
styles.save(chart, path=".", name="pension_by_decile", svg=True)
chart.save("pension_by_decile.png", scale_factor=2.0)
chart
```
