```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

df = pd.DataFrame([
    ("Spring 2026", 23.6),
    ("Now", 4.0),
], columns=["when", "headroom"])
df["lab"] = "£" + df["headroom"].round(0).astype(int).astype(str) + "bn"
df["hi"] = df["when"] == "Now"

bars = alt.Chart(df).mark_bar(cornerRadius=3, size=90).encode(
    x=alt.X("when:N", sort=["Spring 2026","Now"], title=None,
            axis=alt.Axis(labelFontSize=13, labelFontWeight="bold")),
    y=alt.Y("headroom:Q", title="Fiscal headroom (£ billion)",
            scale=alt.Scale(domain=[0, 26])),
    color=alt.Color("hi:N",
        scale=alt.Scale(domain=[True, False], range=["#E6224B", "#36B7B4"]),
        legend=None))

labels = alt.Chart(df).mark_text(dy=-10, fontSize=16, fontWeight="bold").encode(
    x=alt.X("when:N", sort=["Spring 2026","Now"]),
    y="headroom:Q", text="lab:N",
    color=alt.Color("hi:N", scale=alt.Scale(domain=[True, False],
        range=["#E6224B", "#36B7B4"]), legend=None))

# comfortable / danger reference lines
refs = pd.DataFrame([("Comfortable", 20), ("Danger", 10)],
                    columns=["label", "y"])
ref_lines = alt.Chart(refs).mark_rule(color="#676A86", strokeDash=[4,4],
    strokeWidth=1, opacity=0.7).encode(y="y:Q")
ref_txt = alt.Chart(refs).mark_text(align="right", dx=-4, dy=-5, fontSize=10,
    color="#676A86").encode(x=alt.value(600), y="y:Q", text="label:N")

caption = alt.Title(
    text="Source: Ed Conway / Sky News (HM Treasury / OBR figures), September 2026",
    subtitle=[
        "The Chancellor's headroom against the borrowing rule has collapsed from £24bn to about £4bn.",
    ],
    orient="bottom", anchor="start", fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (ref_lines + ref_txt + bars + labels).properties(
    width=700, height=270, title=caption).configure(
    background="white", font="Circular Std").configure_view(
    fill="transparent", stroke="transparent")

styles.save(chart, path=".", name="headroom_simple", svg=True)
chart.save("headroom_simple.png", scale_factor=2.0)
chart
```
