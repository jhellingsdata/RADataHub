```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# The 10 sets of UK fiscal rules since 1997 (Institute for Government, 2025)
sets = pd.DataFrame([
    ("New Labour rules",     1997, 2009, "Labour"),
    ("Darling rules",        2009, 2010, "Labour"),
    ("Coalition I",          2010, 2014, "Conservative"),
    ("Coalition II",         2014, 2015, "Conservative"),
    ("Osborne surplus rule", 2015, 2016, "Conservative"),
    ("Hammond rules",        2016, 2019, "Conservative"),
    ("Javid rules",          2019, 2020, "Conservative"),
    ("Sunak rules",          2021, 2022, "Conservative"),
    ("Hunt rules",           2022, 2024, "Conservative"),
    ("Reeves rules",         2024, 2026, "Labour"),
], columns=["ruleset", "start", "end", "party"])

CON, LAB = "#0087DC", "#E4003B"

chart = alt.Chart(sets).mark_bar(height=16, cornerRadius=2).encode(
    x=alt.X("start:Q", scale=alt.Scale(domain=[1997, 2027], nice=False),
            axis=alt.Axis(format="d", grid=True), title=None),
    x2="end:Q",
    y=alt.Y("ruleset:N", sort=alt.EncodingSortField("start"), title=None),
    color=alt.Color("party:N",
        scale=alt.Scale(domain=["Labour", "Conservative"], range=[LAB, CON]),
        legend=alt.Legend(orient="top", direction="horizontal")),
).properties(width=620, height=alt.Step(26),
    title=alt.Title("The fiscal-rules graveyard",
        subtitle="Each set of UK fiscal rules since 1997, and how long it lasted"))

styles.save(chart, name="fiscal_rules", svg=True)
chart
```
