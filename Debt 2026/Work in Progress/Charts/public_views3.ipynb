```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

df = pd.DataFrame([
    ("Government is in too much debt", 50),
    ("Government has no money",         5),
], columns=["stmt", "net"])
df["series"] = "net"   # constant, so theme assigns one palette colour

base = alt.Chart(df).encode(
    y=alt.Y("stmt:N", sort="-x", title=None,
            axis=alt.Axis(labelFontSize=13, labelLimit=300)))

bars = base.mark_bar(cornerRadius=2).encode(
    x=alt.X("net:Q", title="Net agreement (% agree − % disagree)",
            scale=alt.Scale(domain=[0, 60]),
            axis=alt.Axis(labelFontSize=11, grid=True)),
    color=alt.Color("series:N", legend=None))   # theme colours it

labels = base.mark_text(align="left", dx=6, fontSize=14,
                        color="#676A86", fontWeight="bold").encode(
    x="net:Q", text=alt.Text("net:Q", format="+d"))

caption = alt.Title(
    text="Source: Public First, fieldwork 11–15 March 2025 (n=2,055)",
    subtitle=[
        "Net agreement with two statements about the public finances.",
        "Near-consensus that debt is too high — but no agreement the government is broke.",
    ],
    orient="bottom", anchor="start",
    fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (bars + labels)
    .properties(width=560, height=alt.Step(70), title=caption)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent")
    .configure_axis(labelColor="#676A86", titleColor="#676A86",
                    gridColor="#e5e7eb", domainColor="#cbd5e1"))

styles.save(chart, name="public_views3", svg=True)
chart
```
