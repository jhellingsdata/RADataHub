```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# World Bank NY.GDP.PCAP.KD — Greece real GDP per capita, constant 2015 US$, 1960–2025
vals = [4985.76049516035,5599.4972624979,5586.54223274246,6225.1293347158,6786.24148561981,
    7481.90330099547,7909.24507897853,8289.86933775607,8829.42258970423,9814.49791967552,
    10666.6208776244,11453.2123477243,12535.1276831236,13488.1243950179,12573.3482017471,
    13248.9212623097,13938.5385978162,14162.9910906399,14993.6891417866,15293.9313177537,
    15246.994403616,14876.1172004964,14617.2352230695,14375.6996589942,14591.8630536729,
    14900.0862258833,14927.7613419499,14541.8603092112,15110.414666476,15602.9733207597,
    15438.7937023583,15727.4735434744,15717.04638162,15374.8626265369,15604.0339944242,
    15857.4161930856,16005.9511569711,16501.0316491318,16973.8293714071,17358.4346978294,
    18002.9064613985,18742.4235780691,19548.3677447396,20632.2557041942,21688.1050959558,
    21880.5265835801,23220.5324419197,23973.6936091313,23923.8800731311,22878.1346644397,
    21547.724178239,19448.255802525,17924.6620496886,17644.8766635021,17903.5252440012,
    17980.7298238137,18049.928997892,18352.0884461258,18769.0389621976,19216.6759579785,
    17486.951333323,19231.2888803517,20550.5049158297,21050.8648111581,21494.6856060599,
    21921.6862747184]
years = list(range(1960, 1960 + len(vals)))
df = pd.DataFrame({"year": years, "value": vals})
df = df[df["year"] >= 1995].copy()      # match the GDP-index window
df["date"] = pd.to_datetime(df["year"], format="%Y")

base = df.loc[df["year"] == 2007, "value"].iloc[0]   # index to 2007 peak
df["index"] = df["value"] / base * 100
df["series"] = "pc"

rule_100 = alt.Chart(pd.DataFrame({"y": [100]})).mark_rule(
    color="#cbd5e1", strokeDash=[1, 5], strokeWidth=1).encode(y="y:Q")

line = alt.Chart(df).mark_line(strokeWidth=1.6).encode(
    x=alt.X("date:T", axis=alt.Axis(format="%Y", tickCount=8), title=None),
    y=alt.Y("index:Q", scale=alt.Scale(zero=False),
            title="Real GDP per capita (2007 = 100)"),
    color=alt.Color("series:N", legend=None))

caption = alt.Title(
    text="Source: World Bank (NY.GDP.PCAP.KD)",
    subtitle=[
        "Real GDP per capita, constant 2015 US$, indexed to 2007 = 100. Annual, 1995–2025.",
        "Per person, Greeks are still around 9% poorer than before the crisis.",
    ],
    orient="bottom", anchor="start",
    fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (rule_100 + line)
    .properties(width=700, height=280, title=caption)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent")
    .configure_axis(labelColor="#676A86", titleColor="#676A86"))

styles.save(chart, name="greece_gpc", svg=True)
chart
```
