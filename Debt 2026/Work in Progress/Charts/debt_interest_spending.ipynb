```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# HMRC Annual Tax Summary / PESA Table 5.2, 2024-25. £bn, expenditure on services.
df = pd.DataFrame([
    ("Welfare", 246.1), ("Health", 241.8), ("State pensions", 137.8),
    ("Debt interest", 124.7), ("Education", 118.7), ("Defence", 63.6),
    ("Public order & safety", 51.4), ("Transport", 46.4),
    ("Business & industry", 40.4), ("Government admin", 23.0),
    ("Housing & utilities", 22.3), ("Environment", 17.1),
    ("Culture", 14.5), ("Overseas aid", 7.7),
], columns=["fn", "spend"])
df["hi"] = df["fn"] == "Debt interest"   # highlight the one that matters

base = alt.Chart(df).encode(
    y=alt.Y("fn:N", sort="-x", title=None,
            axis=alt.Axis(labelFontSize=12, labelLimit=200)))

bars = base.mark_bar(cornerRadius=2, size=20).encode(
    x=alt.X("spend:Q", title=None, axis=None, scale=alt.Scale(domain=[0, 260])),
    color=alt.Color("hi:N",
        scale=alt.Scale(domain=[True, False], range=["#E6224B", "#36B7B4"]),
        legend=None))

labels = base.mark_text(align="left", dx=5, fontSize=11, color="#676A86").encode(
    x="spend:Q", text=alt.Text("spend:Q", format=".0f"))

caption = alt.Title(
    text="Source: HMRC Annual Tax Summary / PESA Table 5.2, 2024–25",
    subtitle=[
        "UK public spending by function, £ billion.",
        "Debt interest is the fourth-largest line — above education, nearly double defence.",
    ],
    orient="bottom", anchor="start",
    fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (bars + labels)
    .properties(width=560, height=420, title=caption)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent"))

styles.save(chart, name="debt_interest_spending", svg=True)
chart
```
