```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# UK 10-year gilt yield, Sept–Oct 2022 (LSEG, GB10YT=RR)
rows = """06/09/2022,3.084
07/09/2022,3.009
08/09/2022,3.155
09/09/2022,3.093
12/09/2022,3.069
13/09/2022,3.172
14/09/2022,3.127
15/09/2022,3.161
16/09/2022,3.131
19/09/2022,3.152
20/09/2022,3.275
21/09/2022,3.304
22/09/2022,3.491
23/09/2022,3.819
26/09/2022,4.251
27/09/2022,4.485
28/09/2022,4.036
29/09/2022,4.238
30/09/2022,4.105
03/10/2022,3.966
04/10/2022,3.859
05/10/2022,4.035
06/10/2022,4.192
07/10/2022,4.254
10/10/2022,4.458
11/10/2022,4.423
12/10/2022,4.444
13/10/2022,4.183
14/10/2022,4.378
17/10/2022,3.981
18/10/2022,3.938
19/10/2022,3.893
20/10/2022,3.959
21/10/2022,4.052
24/10/2022,3.714
25/10/2022,3.647"""
data = pd.DataFrame([r.split(",") for r in rows.strip().split("\n")], columns=["date", "yield"])
data["date"] = pd.to_datetime(data["date"], dayfirst=True)
data["yield"] = data["yield"].astype(float)

INK, GREY = "#122b39", "#676A86"

line = alt.Chart(data).mark_line(color=INK, strokeWidth=2.2).encode(
    x=alt.X("date:T", axis=alt.Axis(format="%d %b", tickCount=7), title=None),
    y=alt.Y("yield:Q", title="UK 10-year gilt yield, %", scale=alt.Scale(zero=False)))

events = pd.DataFrame([
    ("2022-09-23", "Mini-budget"),
    ("2022-10-14", "Kwarteng sacked"),
    ("2022-10-20", "Truss resigns"),
], columns=["date", "label"])
events["date"] = pd.to_datetime(events["date"])

rules = alt.Chart(events).mark_rule(color=GREY, strokeDash=[3, 3], strokeWidth=1).encode(x="date:T")

def evt_label(date, text, align, dx, ypix):
    d = events[events["date"] == pd.to_datetime(date)]
    return alt.Chart(d).mark_text(color=GREY, align=align, dx=dx, dy=10,
                                  fontSize=11).encode(
        x="date:T", y=alt.value(ypix), text="label:N")

# stagger heights: mini-budget top-left; Kwarteng higher; Truss a row below & leaning left
labs = (evt_label("2022-09-23", "Mini-budget",     "left",   4,  0)
      + evt_label("2022-10-14", "Kwarteng sacked", "right", -4,  0)
      + evt_label("2022-10-20", "Truss resigns",   "left",   4, 22))

chart = (rules + line + labs).properties(
    width=640, height=340,
    title=alt.Title("The Truss premiership, in one market",
                    subtitle="UK 10-year gilt yield, Sept–Oct 2022"))

styles.save(chart, name="truss_gilts", svg=True)
chart
```
