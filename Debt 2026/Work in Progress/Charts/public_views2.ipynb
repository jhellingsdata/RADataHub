```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# % saying "definitely" or "probably possible", by age. Public First, Mar 2025, n=2,055.
df = pd.DataFrame([
    ("18–24", 63), ("25–34", 62), ("35–44", 49),
    ("45–54", 43), ("55–64", 32), ("65+", 35),
], columns=["age", "possible"])
df["series"] = "possible"          # constant -> theme assigns one palette colour
TOTAL = 47

base = alt.Chart(df).encode(
    x=alt.X("age:N", sort=list(df["age"]), title=None,
            axis=alt.Axis(labelFontSize=12, labelAngle=0)))

bars = base.mark_bar(width=44).encode(
    y=alt.Y("possible:Q", scale=alt.Scale(domain=[0, 70]),
            title="% saying possible",
            axis=alt.Axis(values=[0,10,20,30,40,50,60,70],
                          labelFontSize=11, grid=True)),
    color=alt.Color("series:N", legend=None))     # theme colours it

labels = base.mark_text(dy=-8, fontSize=12, color="#676A86").encode(
    y="possible:Q", text=alt.Text("possible:Q", format="d"))

avg = alt.Chart(pd.DataFrame({"y":[TOTAL]})).mark_rule(
    color="#94a3b8", strokeDash=[4,4], strokeWidth=1).encode(y="y:Q")
avg_txt = alt.Chart(pd.DataFrame({"age":["65+"], "y":[TOTAL], "t":["All adults 47%"]})).mark_text(
    align="right", dx=20, dy=-6, fontSize=10, color="#676A86").encode(
    x=alt.X("age:N", sort=list(df["age"])), y="y:Q", text="t:N")

caption = alt.Title(
    text="Source: Public First, fieldwork 11–15 March 2025 (n=2,055)",
    subtitle=[
        "“Is it possible to reduce taxes and debt while increasing spending?”  % saying possible.",
        "Nearly half think you can have it all — and belief in it falls sharply with age.",
    ],
    orient="bottom", anchor="start",
    fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (bars + labels + avg + avg_txt)
    .properties(width=700, height=280, title=caption)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent")
    .configure_axis(labelColor="#676A86", titleColor="#676A86",
                    gridColor="#e5e7eb", domainColor="#cbd5e1"))

styles.save(chart, name="public_views2", svg=True)
chart
```
