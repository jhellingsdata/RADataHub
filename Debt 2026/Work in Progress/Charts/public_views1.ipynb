```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

df = pd.DataFrame([
    ("Increase health spending",    77),
    ("Reduce the UK debt",          66),
    ("Reduce taxes on individuals", 46),
    ("Increase defence spending",   39),
    ("Reduce taxes on businesses",  13),
], columns=["stmt", "net"])
df["series"] = "net"   # constant -> theme assigns one palette colour

base = alt.Chart(df).encode(
    y=alt.Y("stmt:N", sort="-x", title=None,
            axis=alt.Axis(labelFontSize=12, labelLimit=220)))

bars = base.mark_bar(height=26).encode(
    x=alt.X("net:Q", title=None, axis=None,
            scale=alt.Scale(domain=[0, 85])),
    color=alt.Color("series:N", legend=None))   # theme colours it

labels = base.mark_text(align="left", dx=5, fontSize=12, color="#676A86").encode(
    x="net:Q", text=alt.Text("net:Q", format="+d"))

caption = alt.Title(
    text="Source: Public First, fieldwork 11–15 March 2025",
    subtitle=[
        "“Do you agree the UK government should…”  Net agreement (% agree − % disagree).",
        "Every option is net-positive — including cutting debt and the things that raise it.",
    ],
    orient="bottom", anchor="start",
    fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (bars + labels)
    .properties(width=700, height=alt.Step(46), title=caption)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent"))

styles.save(chart, name="public_views1", svg=True)
chart
```
