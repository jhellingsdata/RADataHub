```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

years = list(range(1991, 2026))

# World Bank — Greece unemployment, % of labour force (modelled ILO)
overall = [7.572, 8.927, 10.177, 9.808, 10.116, 10.402, 10.472, 12.227, 12.138,
    11.345, 10.764, 10.346, 9.839, 10.628, 10.074, 8.906, 8.351, 7.664, 9.546,
    12.719, 17.973, 24.731, 27.686, 26.708, 24.981, 23.514, 21.413, 19.179,
    17.045, 15.899, 14.656, 12.426, 11.023, 10.02, 8.54]           # SL.UEM.TOTL.ZS
youth = [23.858, 26.963, 30.392, 28.926, 30.27, 31.52, 32.096, 32.178, 31.316,
    28.957, 27.816, 26.711, 26.499, 26.57, 25.811, 24.658, 22.731, 21.61, 25.469,
    33.005, 44.951, 55.656, 58.491, 52.645, 49.731, 47.223, 43.392, 40.011,
    34.754, 33.74, 35.39, 31.29, 26.543, 22.375, 21.389]          # SL.UEM.1524.ZS

df = pd.concat([
    pd.DataFrame({"year": years, "rate": youth,   "series": "Youth (15–24)"}),
    pd.DataFrame({"year": years, "rate": overall, "series": "All ages"}),
], ignore_index=True)
df["date"] = pd.to_datetime(df["year"], format="%Y")

# theme assigns colours; youth first in the domain so it takes the primary colour
color = alt.Color("series:N",
    scale=alt.Scale(domain=["Youth (15–24)", "All ages"]),
    legend=None)

lines = alt.Chart(df).mark_line(strokeWidth=1.6).encode(
    x=alt.X("date:T", axis=alt.Axis(format="%Y", tickCount=8), title=None),
    y=alt.Y("rate:Q", scale=alt.Scale(zero=True), title="Unemployment rate (%)"),
    color=color)

labels = alt.Chart(df).mark_text(align="left", dx=6, fontSize=12).encode(
    x=alt.X("date:T", aggregate={"argmax": "date"}),
    y=alt.Y("rate:Q", aggregate={"argmax": "date"}),
    text="series:N", color=color)

caption = alt.Title(
    text="Source: World Bank",
    subtitle=[
        "Unemployment, % of labour force (modelled ILO estimate). Annual, 1991–2025.",
        "Youth unemployment peaked at 58.5% in 2013.",
    ],
    orient="bottom", anchor="start",
    fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (lines + labels)
    .properties(width=700, height=280, title=caption)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent")
    .configure_axis(labelColor="#676A86", titleColor="#676A86"))

styles.save(chart, name="greece_unemployment", svg=True)
chart
```
