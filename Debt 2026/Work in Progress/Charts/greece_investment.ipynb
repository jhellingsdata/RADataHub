```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# World Bank NE.GDI.FTOT.ZS — Greece gross fixed capital formation, % of GDP, 1960–2025
vals = [23.607073965236,23.3175152995366,25.5420600602326,22.2636222958609,24.2895975739071,
25.2321188352563,26.2761333501366,25.1384618441379,28.6347167588621,30.0377219286972,
29.0172299175777,30.3142831193755,34.8593217030721,35.1978598623406,27.022978907553,
26.9074115841207,27.5132275132275,31.2910955174184,34.84004196387,37.2040572538979,
32.2086506960762,29.3813022292793,26.5953792620779,28.4995756574332,23.0503846696489,
24.7603967764711,25.7438752071109,24.3511104389388,24.2671136606532,25.4089069966259,
26.0396039118894,25.4252437850636,24.0030936575419,22.8334881362765,21.0129577300315,
20.9666910176002,22.1084202126939,21.5899391374926,24.5135902861657,25.3860319397874,
25.9373547837658,25.5981996968449,23.875854552048,24.8780068243839,25.0449225325719,
21.3568870584015,24.0565849914051,25.5308773380251,23.2806962442851,20.438446980754,
16.8200572069043,13.8655955400985,12.2795312839599,11.5362301011246,11.233781428641,
11.1610941969144,11.3684708155605,12.0492834143889,11.3310526081969,10.97046317576,
12.2998658740322,13.7915466530779,15.7213358540523,15.8842464967092,16.0395231559948,
16.881202808615]
years = list(range(1960, 1960 + len(vals)))
df = pd.DataFrame({"year": years, "gfcf": vals})
df = df[df["year"] >= 1995].copy()
df["date"] = pd.to_datetime(df["year"], format="%Y")
df["series"] = "gfcf"

line = alt.Chart(df).mark_line(strokeWidth=1.8).encode(
    x=alt.X("date:T", axis=alt.Axis(format="%Y", tickCount=7), title=None),
    y=alt.Y("gfcf:Q", scale=alt.Scale(domain=[10, 26]),
            title="Investment (% of GDP)",
            axis=alt.Axis(labelExpr="datum.label + '%'")),
    color=alt.Color("series:N", legend=None))

# mark crisis onset
crisis = alt.Chart(pd.DataFrame({"date":pd.to_datetime(["2009-01-01"])})).mark_rule(
    color="#94a3b8", strokeDash=[2,3]).encode(x="date:T")
crisis_txt = alt.Chart(pd.DataFrame({"date":pd.to_datetime(["2009-01-01"]),
    "t":["Crisis"]})).mark_text(align="left", dx=4, dy=-4, fontSize=10,
    color="#64748b").encode(x="date:T", y=alt.value(12), text="t:N")

caption = alt.Title(
    text="Source: World Bank (NE.GDI.FTOT.ZS)",
    subtitle=[
        "Greek gross fixed capital formation (investment), % of GDP. Annual, 1995–2025.",
        "Investment more than halved after 2009 — the debt overhang acting as a tax on investment.",
    ],
    orient="bottom", anchor="start", fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (crisis + crisis_txt + line)
    .properties(width=700, height=270, title=caption)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent"))

styles.save(chart, name="greece_investment", svg=True)
chart
```
