```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# Greek gross fixed capital formation, volume (€m, quarterly). Source: ECB (MNA series).
df = pd.read_csv("../Data/ECB Data Portal_20260923170915.csv")
df.columns = ["date", "q", "gfcf"]
df["date"] = pd.to_datetime(df["date"])
df["series"] = "x"

line = alt.Chart(df).mark_line(strokeWidth=1.8).encode(
    x=alt.X("date:T", axis=alt.Axis(format="%Y", tickCount=8), title=None),
    y=alt.Y("gfcf:Q", scale=alt.Scale(domain=[0, 18000]),
            title="Investment (€m per quarter, volume)"),
    color=alt.Color("series:N", legend=None))

# crisis marker
crisis = alt.Chart(pd.DataFrame({"date":pd.to_datetime(["2009-01-01"])})).mark_rule(
    color="#94a3b8", strokeDash=[2,3]).encode(x="date:T")
crisis_txt = alt.Chart(pd.DataFrame({"date":pd.to_datetime(["2009-01-01"]),
    "t":["Crisis"]})).mark_text(align="left", dx=4, dy=-4, fontSize=10,
    color="#64748b").encode(x="date:T", y=alt.value(12), text="t:N")

caption = alt.Title(
    text="Source: ECB (gross fixed capital formation, volume)",
    subtitle=[
        "Greek investment, € millions per quarter. Quarterly, 1995–2026.",
    ],
    orient="bottom", anchor="start", fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (crisis + crisis_txt + line)
    .properties(width=700, height=300, title=caption)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent"))

styles.save(chart, path=".", name="greece_investment", svg=True)
chart.save("greece_investment.png", scale_factor=2.0)
chart
```
