```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# OBR July 2026 Fiscal Risks and Sustainability report, Chart 5.3 (exact annual data).
fy = list(range(2025, 2076))  # fiscal-year start; 2025 = 2025-26 ... 2075 = 2075-76
baseline = [94.3,94.8,95.7,96.3,96.1,95.1,94.8,94.8,95.4,96.4,97.8,99.1,100.7,102.2,
103.9,105.8,107.8,109.9,112.2,114.6,117.1,119.8,122.7,125.9,129.2,132.8,136.5,140.5,
144.7,149.1,153.9,158.9,164.1,169.6,175.3,181.2,187.4,193.9,200.7,207.7,214.9,222.4,
230.1,237.9,246.1,254.4,263.1,272.2,281.4,290.7,300.4]
shocks = [94.3,96.1,97.9,99.8,101.5,103.1,105.4,108.3,111.7,115.8,130.3,134.9,139.6,
144.5,149.7,155.1,160.7,166.4,172.4,188.5,194.8,201.3,208.0,215.0,222.3,229.8,237.5,
245.5,263.8,272.4,281.2,290.4,299.9,309.6,319.6,329.8,340.3,361.2,372.4,383.8,395.5,
407.4,419.6,431.8,444.3,457.1,480.2,493.7,507.3,521.2,535.6]

rows = []
for y, b, s in zip(fy, baseline, shocks):
    rows.append((y, b, "Baseline"))
    rows.append((y, s, "With repeated shocks"))
df = pd.DataFrame(rows, columns=["year", "pct", "scenario"])
df["date"] = pd.to_datetime(df["year"], format="%Y")

color = alt.Color("scenario:N", legend=None)

lines = alt.Chart(df).mark_line(strokeWidth=2.4).encode(
    x=alt.X("date:T", axis=alt.Axis(format="%Y", tickCount=7), title=None),
    y=alt.Y("pct:Q", scale=alt.Scale(domain=[0, 560]),
            title="Per cent of GDP",
            axis=alt.Axis(labelExpr="datum.label + '%'")),
    color=color, detail="scenario:N")

labels = alt.Chart(df).mark_text(align="left", dx=6, fontSize=12, fontWeight="bold").encode(
    x=alt.X("date:T", aggregate="max"),
    y=alt.Y("pct:Q", aggregate={"argmax":"date"}),
    text=alt.Text("scenario:N", aggregate={"argmax":"date"}),
    color=color)

caption = alt.Title(
    text="Source: OBR, July 2026 Fiscal Risks and Sustainability report",
    subtitle=[
        "Long-term projection of UK public sector net debt, % of GDP, on unchanged policy.",
    ],
    orient="bottom", anchor="start", fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (lines + labels)
    .properties(width=700, height=270, title=caption)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent"))

styles.save(chart, path=".", name="obr_longterm_projection", svg=True)
chart.save("obr_longterm_projection.png", scale_factor=2.0)
chart
```
