```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# OBR March 2026 (Chart 4.6): debt interest as % of government revenue
rev = [9.7,9.7,9.5,9.8,10.0,9.5,8.9,8.9,8.0,7.3,7.4,7.4,6.8,6.2,5.8,6.2,6.5,6.7,6.2,
6.4,7.1,7.7,8.2,8.8,8.9,9.2,9.1,9.4,9.7,10.2,10.2,10.1,9.5,9.2,8.6,7.5,8.0,8.4,8.8,
9.2,9.4,8.9,8.3,6.8,6.6,5.6,5.2,5.2,5.4,5.3,5.6,5.8,5.9,5.0,6.8,7.0,6.1,5.7,5.0,4.9,
4.9,5.5,4.8,4.7,3.1,5.9,10.8,9.8,9.3,8.9,8.4,8.5,8.8,9.0,8.8]
start = 1956
df = pd.DataFrame({"fy_start": list(range(start, start+len(rev))), "pct": rev})
df["date"] = pd.to_datetime(df["fy_start"], format="%Y")
FC = 2025
df["seg"] = df["fy_start"].apply(lambda y: "forecast" if y >= FC else "actual")

base = alt.Chart(df).encode(
    x=alt.X("date:T", axis=alt.Axis(format="%Y", tickCount=8), title=None),
    y=alt.Y("pct:Q", scale=alt.Scale(domain=[0, 12]),
            title="Debt interest (% of government revenue)",
            axis=alt.Axis(labelExpr="datum.label + '%'")))

actual = base.transform_filter("datum.seg=='actual'").mark_line(
    strokeWidth=2, color="#E6224B")
fc = base.transform_filter("datum.fy_start>=2024").mark_line(
    strokeWidth=2, color="#E6224B", strokeDash=[4,3])

caption = alt.Title(
    text="Source: OBR, Economic and Fiscal Outlook, March 2026 (Chart 4.6)",
    subtitle=[
        "UK debt interest as a share of government revenue, %. Dashed = forecast.",
        "Nearly a tenth of every tax pound now services debt — back to 1980s levels.",
    ],
    orient="bottom", anchor="start", fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (actual + fc)
    .properties(
        width=700, height=270, title=caption,
        autosize=alt.AutoSizeParams(type="fit", contains="padding"),  # <- was missing
    )
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent"))

styles.save(chart, path=".", name="debt_interest_revenue", svg=True)
# crisp raster for the blog:
chart.save("debt_interest_revenue.png", scale_factor=3)   # 2100×810, same 700×270 ratio
chart
```
