```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# "What do you think the government considers most?" (select up to 3). All adults, Mar 2025, n=2,055.
df = pd.DataFrame([
    ("Winning the next election",   44),
    ("What their party donors want",35),
    ("What's best for the country", 34),
    ("What they personally want",   32),
    ("What's best long term",       30),
    ("Expert advice",               20),
    ("What the public want",        19),
    ("What's best short term",      19),
    ("What's best for the world",   12),
], columns=["motive", "pct"])

# highlight the two cynical top answers
df["hi"] = df["motive"].isin(["Winning the next election", "What their party donors want"])

base = alt.Chart(df).encode(
    y=alt.Y("motive:N", sort="-x", title=None,
            axis=alt.Axis(labelFontSize=12, labelLimit=200)))

bars = base.mark_bar(height=20, cornerRadius=2).encode(
    x=alt.X("pct:Q", title=None, axis=alt.Axis(format="d", grid=True),
            scale=alt.Scale(domain=[0, 50])),
    color=alt.Color("hi:N",
        scale=alt.Scale(domain=[True, False], range=["#2563eb", "#cbd5e1"]),
        legend=None))

labels = base.mark_text(align="left", dx=5, fontSize=11, color="#676A86").encode(
    x="pct:Q", text=alt.Text("pct:Q", format="d"))

caption = alt.Title(
    text="Source: Public First, fieldwork 11–15 March 2025 (n=2,055)",
    subtitle=[
        "“What do you think the government considers most on the economy?”  Select up to three.",
        "Voters think elections and donors drive economic policy — above the national interest.",
    ],
    orient="bottom", anchor="start",
    fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (bars + labels)
    .properties(width=480, height=alt.Step(30), title=caption)
    .configure(background="#f7f7f7", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent")
    .configure_axis(labelColor="#676A86", titleColor="#676A86",
                    gridColor="#e5e7eb", domainColor="#cbd5e1"))

styles.save(chart, name="public_views4", svg=True)
chart
```
