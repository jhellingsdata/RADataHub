```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

df = pd.read_csv("../Data/time_series_GB_20040101-0000_20260911-1645.csv")
df.columns = ["date", "value"]
df["date"] = pd.to_datetime(df["date"])
df["value"] = pd.to_numeric(df["value"], errors="coerce")

line = alt.Chart(df).mark_line(color="#e6224b", strokeWidth=1.8).encode(
    x=alt.X("date:T", axis=alt.Axis(format="%Y", tickCount=8), title=None),
    y=alt.Y("value:Q", title="Search interest (0–100)"))

# Source + explainer live in a bottom-oriented title (Altair has no native caption slot)
caption = alt.Title(
    text="Source: Google Trends",
    subtitle=[
        "Values are relative, not absolute search volumes: 100 marks the single highest point of",
        "search interest for the term over 2004–2026; 50 means half that level, and so on.",
    ],
    orient="bottom", anchor="start",
    fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86",
    dy=12,
)

chart = line.properties(width=400, height=300, title=caption) \
    .configure(background="#f7f7f7", font="Circular Std") \
    .configure_view(fill="transparent", stroke="transparent") \
    .configure_axis(labelColor="#676A86", titleColor="#676A86")

styles.save(chart, name="gov_debt_searches", svg=True)
chart
```
