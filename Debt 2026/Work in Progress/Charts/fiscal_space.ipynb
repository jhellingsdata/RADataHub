```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# UK net debt (% GDP) at the start of each major shock. Source: ONS PUSF/HF6X.
df = pd.DataFrame([
    ("Financial crisis\n2008", 36.1),
    ("COVID\n2020",            84.2),
    ("Next shock?\ntoday",     94.6),
], columns=["shock", "debt"])
df["hi"] = df["shock"].str.contains("Next")   # highlight 'today'

base = alt.Chart(df).encode(
    x=alt.X("shock:N", sort=list(df["shock"]), title=None,
            axis=alt.Axis(labelFontSize=12, labelAngle=0)))

bars = base.mark_bar(cornerRadius=2, size=90).encode(
    y=alt.Y("debt:Q", scale=alt.Scale(domain=[0, 100]),
            title="Net debt entering the shock (% of GDP)"),
    color=alt.Color("hi:N",
        scale=alt.Scale(domain=[True, False], range=["#E6224B", "#36B7B4"]),
        legend=None))

labels = base.mark_text(dy=-10, fontSize=14, fontWeight="bold", color="#676A86").encode(
    y="debt:Q", text=alt.Text("debt:Q", format=".0f"))

caption = alt.Title(
    text="Source: ONS (PUSF / HF6X)",
    subtitle=[
        "UK public sector net debt, % of GDP, at the onset of each major shock.",
        "Each crisis starts from a higher base — leaving less room to respond to the next.",
    ],
    orient="bottom", anchor="start",
    fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (
    (bars + labels)
    .properties(width=520, height=340, title=caption)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent"))

styles.save(chart, name="fiscal_space", svg=True)
chart
```
