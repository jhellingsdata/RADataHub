```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

# OBR March 2026 (Chart 4.6): debt interest as % of government revenue
rev = [9.7,9.7,9.5,9.8,10.0,9.5,8.9,8.9,8.0,7.3,7.4,7.4,6.8,6.2,5.8,6.2,6.5,6.7,6.2,
6.4,7.1,7.7,8.2,8.8,8.9,9.2,9.1,9.4,9.7,10.2,10.2,10.1,9.5,9.2,8.6,7.5,8.0,8.4,8.8,
9.2,9.4,8.9,8.3,6.8,6.6,5.6,5.2,5.2,5.4,5.3,5.6,5.8,5.9,5.0,6.8,7.0,6.1,5.7,5.0,4.9,
4.9,5.5,4.8,4.7,3.1,5.9,10.8,9.8,9.3,8.9,8.4,8.5,8.8,9.0,8.8]
start = 1956
df = pd.DataFrame({"fy_start": list(range(start, start+len(rev))), "pct": rev})
df["date"] = pd.to_datetime(df["fy_start"], format="%Y")
FC = 2025
df["seg"] = df["fy_start"].apply(lambda y: "forecast" if y >= FC else "actual")

# for the tooltip: financial-year label and the basis of each number
df["fy"] = df["fy_start"].apply(lambda y: f"{y}–{str(y + 1)[-2:]}")
df["basis"] = df["seg"].map({"actual": "Outturn", "forecast": "OBR forecast"})

base = alt.Chart(df).encode(
    x=alt.X("date:T", axis=alt.Axis(format="%Y", tickCount=8), title=None),
    y=alt.Y("pct:Q", scale=alt.Scale(domain=[0, 12]),
            title="Debt interest (% of government revenue)",
            axis=alt.Axis(labelExpr="datum.label + '%'")))

actual = base.transform_filter("datum.seg=='actual'").mark_line(
    strokeWidth=2, color="#E6224B")
fc = base.transform_filter("datum.fy_start>=2024").mark_line(
    strokeWidth=2, color="#E6224B", strokeDash=[4,3])

# --- hover (HTML version only; hidden until you hover, so the static PNG/SVG is unchanged) ---
H = 300
HALF_YEAR = pd.Timedelta(days=182.5)
# one hover cell per year; the first and last stop at their own point so the x-axis range is unchanged
cells_df = df.copy()
cells_df["cell0"] = cells_df["date"] - HALF_YEAR
cells_df["cell1"] = cells_df["date"] + HALF_YEAR
cells_df.loc[cells_df.index[0], "cell0"] = cells_df["date"].iloc[0]
cells_df.loc[cells_df.index[-1], "cell1"] = cells_df["date"].iloc[-1]

hover = alt.selection_point(fields=["date"], on="mouseover", clear="mouseout", empty=False)

cells = (alt.Chart(cells_df).mark_rect(opacity=0)
    .encode(x="cell0:T", x2="cell1:T", y=alt.value(0), y2=alt.value(H),
            tooltip=[alt.Tooltip("fy:N", title="Financial year"),
                     alt.Tooltip("pct:Q", title="Debt interest (% of revenue)", format=".1f"),
                     alt.Tooltip("basis:N", title="Basis")])
    .add_params(hover))

guide = (alt.Chart(df).mark_rule(color="#676A86", strokeWidth=1)
         .encode(x="date:T").transform_filter(hover))
dot = (alt.Chart(df).mark_circle(size=60, color="#E6224B", stroke="white", strokeWidth=1.5, opacity=1)
       .encode(x="date:T", y="pct:Q").transform_filter(hover))

chart = (
    (guide + actual + fc + dot + cells)        # cells last, so they sit on top
    .properties(
        width=400, height=H,
        autosize=alt.AutoSizeParams(type="fit", contains="padding"),  # <- was missing
    )
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent"))

chart.save("debt_interest_revenue.html")                                  # interactive
styles.save(chart, path=".", name="debt_interest_revenue", svg=True)       # static (hover layers hidden)
# crisp raster for the blog:
chart.save("debt_interest_revenue.png", scale_factor=3)   # 2100×810, same 700×270 ratio
chart
```
