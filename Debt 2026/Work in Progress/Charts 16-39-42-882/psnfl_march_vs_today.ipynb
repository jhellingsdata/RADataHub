```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

data = [
    ("2026/27", 82.6, 83.2),
    ("2027/28", 82.9, 83.8),
    ("2028/29", 82.9, 83.8),
    ("2029/30", 82.2, 83.1),
    ("2030/31", 81.1, 82.3),
]
rows = []
for fy, mar, oct_v in data:
    if mar is not None:
        rows.append((fy, mar, "March"))
    rows.append((fy, oct_v, "Today"))
df = pd.DataFrame(rows, columns=["fy", "pct", "vintage"])

order = ["2026/27","2027/28","2028/29","2029/30","2030/31","2031/32"]
voff = ["March","Today"]

# ecostyles palette: teal + red (first two categories)
colour = alt.Color("vintage:N", sort=voff,
    scale=alt.Scale(domain=voff, range=["#36B7B4", "#E6224B"]),
    legend=alt.Legend(orient="top", title=None))

base = alt.Chart(df).encode(
    x=alt.X("fy:N", sort=order, title=None, axis=alt.Axis(labelAngle=0)),
    xOffset=alt.XOffset("vintage:N", sort=voff))

bars = base.mark_bar().encode(
    y=alt.Y("pct:Q", title="Non-financial liabilities (% of GDP)",
            scale=alt.Scale(domain=[80, 85], clamp=True)),
    color=colour)

labels = base.mark_text(align="center", baseline="bottom", dy=-4,
                        fontSize=10, fontWeight="bold", color="#122b39").encode(
    y="pct:Q", text=alt.Text("pct:Q", format=".1f"))

chart = (
    (bars + labels)
    .properties(width=700, height=270)
    .configure(background="white", font="Circular Std")
    .configure_view(fill="transparent", stroke="transparent")
)

styles.save(chart, path=".", name="psnfl_march_vs_today", svg=True)
chart.save("psnfl_march_vs_today.png", scale_factor=2.0)
chart
```
