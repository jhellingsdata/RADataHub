```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

scatter_df = pd.DataFrame([
    ("Japan",          204.4, 2.72),
    ("Italy",          138.4, 4.17),
    ("United States",  125.8, 4.78),
    ("France",         118.4, 4.22),
    ("Canada",         110.7, 3.78),
    ("United Kingdom", 103.6, 5.15),
    ("Germany",         64.6, 3.36),
], columns=["country", "debt", "yield"])

# which side each label sits, so it hugs the dot without overlapping neighbours
label_side = {
    "Japan":          "left",
    "Italy":          "right",
    "United States":  "right",
    "France":         "left",
    "Canada":         "left",
    "United Kingdom": "left",
    "Germany":        "right",
}
scatter_df["al"] = scatter_df.country.map(label_side)
scatter_df
```

```python
styles = EcoStyles(); styles.register_and_enable_theme()
TEAL, RED = "#36b7b4", "#e6224b"
HL = "United Kingdom"

points = (
    alt.Chart(scatter_df)
    .mark_point(filled=True, size=130, opacity=0.9)
    .encode(
        x=alt.X("debt:Q", title="Gross debt (% of GDP)",
        scale=alt.Scale(domain=[50, 220]),
        axis=alt.Axis(labelExpr="datum.label + '%'", tickCount=5,
                      zindex=1, grid=False)),
        y=alt.Y("yield:Q", title="10-year yield (%)",
                scale=alt.Scale(domain=[2, 5.5], nice=False),
                axis=alt.Axis(titleAngle=0, titleAlign="left",
                              titleX=0, titleY=-8, zindex=1)),
        color=alt.condition(alt.datum.country == HL, alt.value(RED), alt.value(TEAL)),
        tooltip=["country", "debt", "yield"],
    )
)

def tl(df, al, dx):
    return (alt.Chart(df).mark_text(fontSize=12, fontWeight="bold", align=al, dx=dx)
            .encode(x="debt:Q", y="yield:Q", text="country:N"))


# labels anchored to the DOT, nudged just off it (12px)
labels = (tl(scatter_df[scatter_df.al == "left"],  "right", -12)
        + tl(scatter_df[scatter_df.al == "left"].iloc[0:0], "left", 12))  # placeholder
labels = (tl(scatter_df[scatter_df.al == "right"], "left",  12)
        + tl(scatter_df[scatter_df.al == "left"],  "right", -12))

chart = (points + labels).properties(
    width=560, height=400,
    title=alt.Title("Gross debt vs 10-year bond yield, G7"),)

styles.save(chart, name="g7_debt_yield", svg=True)
chart
```
