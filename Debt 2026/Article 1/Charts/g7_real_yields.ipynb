```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles(); styles.register_and_enable_theme()

data = pd.DataFrame([
    ("United Kingdom", 5.40, 3.1),
    ("United States",  5.27, 3.4),
    ("France",         4.92, 2.4),
    ("Italy",          4.71, 4.2),
    ("Canada",         3.97, 3.0),
    ("Germany",        3.52, 3.3),
    ("Japan",          3.10, 1.9),
], columns=["country", "Nominal", "cpi"])
data["Real"] = data["Nominal"] - data["cpi"]

long = data.melt("country", value_vars=["Nominal", "Real"],
                 var_name="type", value_name="yield")
order = data.sort_values("Nominal", ascending=False)["country"].tolist()

bars = alt.Chart(long).mark_bar().encode(
    y=alt.Y("country:N", sort=order, title=None,
            axis=alt.Axis(labelFontSize=12, labelLimit=160)),
    yOffset=alt.YOffset("type:N", sort=["Nominal", "Real"]),
    x=alt.X("yield:Q", title="10-year yield, %", scale=alt.Scale(domain=[0, 6])),
    color=alt.Color("type:N", sort=["Nominal", "Real"],
        scale=alt.Scale(domain=["Nominal", "Real"], range=["#36B7B4", "#E6224B"]),
        legend=alt.Legend(orient="top", title=None, labelFontSize=12)))

labels = alt.Chart(long).mark_text(align="left", dx=4, fontSize=10, color="#676A86").encode(
    y=alt.Y("country:N", sort=order),
    yOffset=alt.YOffset("type:N", sort=["Nominal", "Real"]),
    x="yield:Q", text=alt.Text("yield:Q", format=".1f"))

caption = alt.Title(
    text="Source: LSEG (yields); Trading Economics (national CPI). Real ≈ nominal − latest annual inflation.",
    subtitle=[
        "Nominal and real 10-year government bond yields across the G7, %.",
        "The UK has the highest nominal yield — but a big gap to its real yield, because UK inflation is high.",
    ],
    orient="bottom", anchor="start", fontSize=11, subtitleFontSize=10,
    color="#676A86", subtitleColor="#676A86", dy=12)

chart = (bars + labels).properties(width=640, height=360, title=caption).configure(
    background="white", font="Circular Std").configure_view(
    fill="transparent", stroke="transparent")

styles.save(chart, path=".", name="g7_real_yields", svg=True)
chart.save("g7_real_yields.png", scale_factor=2.0)
chart
```
