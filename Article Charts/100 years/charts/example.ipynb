```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

# schedule breakpoints (year, effective death-duty %); step-after carries each rate forward
# eyeballed off the figure — swap in real Cummins values as available
breakpoints = {
    "£5,000":       [(1892,2),(1919,3),(1946,3),(1949,0),(2015,0)],
    "£100,000":     [(1892,5),(1914,6),(1919,7),(1930,7),(1946,6),(1950,3),(1965,2),
                     (1974,9),(1976,3),(1979,10),(1981,2),(1982,0),(2015,0)],
    "£500,000":     [(1892,5),(1914,6),(1919,10),(1930,13),(1940,17),(1946,22),(1950,27),
                     (1960,37),(1968,50),(1972,60),(1974,60),(1976,47),(1978,58),(1980,52),
                     (1982,45),(1985,0),(2015,0)],
    "£1,000,000":   [(1892,6),(1914,7),(1919,13),(1930,17),(1940,25),(1946,33),(1950,40),
                     (1960,52),(1968,63),(1972,70),(1975,70),(1978,68),(1980,70),(1984,68),
                     (1986,40),(2015,40)],
    "£5,000,000":   [(1892,7),(1914,8),(1919,17),(1930,22),(1940,33),(1946,45),(1950,52),
                     (1960,63),(1968,68),(1972,70),(1976,63),(1979,70),(1982,68),(1985,40),
                     (2015,40)],
    "£10,000,000":  [(1892,8),(1914,10),(1919,22),(1925,25),(1930,27),(1940,37),(1946,52),
                     (1950,58),(1960,68),(1968,72),(1972,75),(1975,78),(1978,72),(1980,75),
                     (1983,75),(1985,40),(2015,40)],
    "£100,000,000": [(1892,10),(1914,10),(1919,40),(1925,47),(1928,50),(1934,55),(1938,58),
                     (1940,78),(1946,80),(1968,80),(1974,85),(1976,82),(1978,80),(1980,75),
                     (1983,75),(1985,40),(2015,40)],
}

df = pd.DataFrame(
    [{"year": y, "wealth": w, "rate": r} for w, pts in breakpoints.items() for y, r in pts]
)
df["year"] = pd.to_datetime(df["year"], format="%Y")

# blue (eco primary) for the lowest wealth level ramping to red for the highest
order   = list(breakpoints)
palette = ["#2563eb", "#4fb783", "#a8c34a", "#e6cf4a",
           "#e6a94b", "#e8843c", "#c0392b"]

styles = EcoStyles()
styles.register_and_enable_theme()

chart = (
    alt.Chart(df)
    .mark_line(strokeWidth=1.8, interpolate="step-after")   # duties are step schedules
    .encode(
        x=alt.X("year:T", title=None,
                axis=alt.Axis(format="%Y",
                              values=[f"{y}-01-01" for y in (1890, 1920, 1950, 1980, 2010)])),
        y=alt.Y("rate:Q", title="Death Duty, %",
                scale=alt.Scale(domain=[0, 100]),
                axis=alt.Axis(values=[0, 25, 50, 75, 100])),
        color=alt.Color("wealth:N", sort=order,
                        scale=alt.Scale(domain=order, range=palette),
                        legend=alt.Legend(title="Real Wealth", orient="none",
                                          legendX=70, legendY=40)),
    )
    .properties(width=560, height=380,
                title=alt.Title("Figure 2: Death duties, 1892 to 2015"))
)
styles.save(chart, name="death_duties", svg=True)
chart
```
