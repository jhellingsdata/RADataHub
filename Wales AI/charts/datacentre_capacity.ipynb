```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles()
styles.register_and_enable_theme()

# McKinsey global data centre capacity demand ('continued momentum' scenario), GW.
# AI and non-AI workloads, 2025–2030.
rows = [
    (2025, "AI workload", 44), (2025, "Non-AI workload", 38),
    (2026, "AI workload", 62), (2026, "Non-AI workload", 40),
    (2027, "AI workload", 83), (2027, "Non-AI workload", 45),
    (2028, "AI workload", 102), (2028, "Non-AI workload", 50),
    (2029, "AI workload", 124), (2029, "Non-AI workload", 56),
    (2030, "AI workload", 156), (2030, "Non-AI workload", 64),
]
df = pd.DataFrame(rows, columns=["year", "segment", "gw"])

bars = alt.Chart(df).transform_calculate(
    segment_order="datum.segment === 'AI workload' ? 0 : 1"
).mark_bar().encode(
    x=alt.X("year:O", title=None),
    y=alt.Y("gw:Q", title="Capacity demand, GW", stack="zero"),
    color=alt.Color("segment:N",
        scale=alt.Scale(domain=["AI workload", "Non-AI workload"]),
        legend=alt.Legend(title=None, orient="bottom")),
    order=alt.Order("segment_order:Q"),
)

chart = bars.properties(
    width=560, height=320,
    padding={"bottom": 20, "left": 5, "right": 5, "top": 5},
    title=alt.Title(
        text="Global data centre capacity demand could almost triple by 2030",
        anchor="start",
        subtitle="Source: McKinsey & Company (2025)",
        subtitleColor="#676A86"),
)

styles.save(chart, path=".", name="datacentre_capacity", svg=True)
chart.save("datacentre_capacity.png", scale_factor=2)
chart
```
