```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles()
styles.register_and_enable_theme()

RED, GREY = "#E6224B", "#94a3b8"   # same pair as the brain-drain chart

# Gross domestic expenditure on R&D as % of GDP, 2022.
df = pd.DataFrame({
    "place": ["Israel", "Finland", "UK", "Wales", "Singapore", "Estonia"],
    "pct":   [6.18, 2.98, 2.75, 1.87, 1.81, 1.76],
})
df["hl"] = df["place"].eq("Wales")
df["label"] = df["pct"].map(lambda v: f"{v:.2f}")

# top -> bottom: smallest at top, Israel (largest) at bottom
order = df.sort_values("pct")["place"].tolist()

base = alt.Chart(df).encode(
    y=alt.Y("place:N", title=None, scale=alt.Scale(domain=order)),
    x=alt.X("pct:Q", title="R&D spending, % of GDP", scale=alt.Scale(domain=[0, 7])),
)

bars = base.mark_bar().encode(
    color=alt.Color("hl:N",
        scale=alt.Scale(domain=[True, False], range=[RED, GREY]), legend=None))

labels = base.mark_text(dx=5, align="left", fontWeight="bold").encode(text="label:N")

chart = (bars + labels).properties(
    width=560, height=280,
    title=alt.Title(
        text="Research and development spending, % of GDP, 2022",
        anchor="start",
        subtitle="Source: World Bank; Welsh Government",
        subtitleColor="#676A86",
    ),
)

styles.save(chart, path=".", name="rd_intensity", svg=True)
chart.save("rd_intensity.png", scale_factor=2)
chart
```
