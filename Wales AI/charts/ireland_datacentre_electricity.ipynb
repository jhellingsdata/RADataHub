```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles()
styles.register_and_enable_theme()

# Data centres' share of Ireland's metered electricity consumption, 2015–2025.
# Source: Central Statistics Office (Ireland), Data Centres Metered Electricity
# Consumption 2025.
years = list(range(2015, 2026))
share = [5, 6, 7, 8, 9, 11, 14, 18, 21, 22, 23]
df = pd.DataFrame({"year": pd.to_datetime(years, format="%Y"), "pct": share})

line = alt.Chart(df).mark_line(strokeWidth=2.4).encode(
    x=alt.X("year:T", axis=alt.Axis(format="%Y", tickCount=6), title=None),
    y=alt.Y("pct:Q", title="% of metered electricity", scale=alt.Scale(domain=[0, 25])),
)

# label the final point only
end = df[df.year == pd.Timestamp("2025")].assign(lab=lambda d: d.pct.map(lambda v: f"{v}%"))
labels = alt.Chart(end).mark_text(dy=-10, fontWeight="bold").encode(
    x="year:T", y="pct:Q", text="lab")


chart = (line + labels).properties(
    width=600, height=300,
    title=alt.Title(
        text="Data centres' share of Ireland's metered electricity, 2015–2025",
        anchor="start",
        subtitle="Source: Central Statistics Office (Ireland)",
        subtitleColor="#676A86"),
)

styles.save(chart, path=".", name="ireland_datacentre_electricity", svg=True)
chart.save("ireland_datacentre_electricity.png", scale_factor=2)
chart
```
