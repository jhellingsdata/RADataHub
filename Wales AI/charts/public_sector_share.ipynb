```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles()
styles.register_and_enable_theme()

# Public sector as % of all in employment (ONS APS), Wales vs UK, 2005–2025.
# Values read off gov.wales Figure 6 (the Welsh APS series is not separately
# published for confidentiality, so these are transcribed from the figure).
# CHECK against the figure before use. Endpoints are firm: 2025 Wales 30.5, UK 23.6.
years = list(range(2005, 2026))
wales = [28.3, 29.3, 30.1, 30.0, 30.4, 31.0, 31.5, 30.6, 30.3, 30.2, 29.0,
         27.6, 27.5, 27.0, 26.6, 26.5, 28.3, 29.6, 30.6, 31.5, 30.8]
uk    = [24.0, 24.1, 24.0, 23.9, 24.2, 24.4, 24.5, 23.8, 23.3, 23.0, 22.6,
         22.0, 21.7, 21.5, 21.5, 21.6, 22.9, 23.2, 23.6, 23.9, 23.7]

df = pd.concat([
    pd.DataFrame({"year": years, "pct": wales, "place": "Wales"}),
    pd.DataFrame({"year": years, "pct": uk,    "place": "UK"}),
], ignore_index=True)

color = alt.Color("place:N",
    scale=alt.Scale(domain=["Wales", "UK"]), legend=None)

lines = alt.Chart(df).mark_line(strokeWidth=2.2).encode(
    x=alt.X("year:O", title=None,
            axis=alt.Axis(values=[2005, 2010, 2015, 2020, 2025], labelAngle=0)),
    y=alt.Y("pct:Q", title="Public sector, % of total employment",
            scale=alt.Scale(domain=[0, 35])),
    color=color, detail="place:N")

labels = alt.Chart(df).transform_filter(alt.datum.year == 2025).mark_text(
    align="left", dx=6, fontWeight="bold").encode(
    x="year:O", y="pct:Q", text="place:N", color=color)

chart = (lines + labels).properties(
    width=600, height=320,
    title=alt.Title(
        text="Public sector as a share of total employment, Wales vs UK, 2005–2025",
        anchor="start",
        subtitle="Source: ONS Annual Population Survey (APS)",
        subtitleColor="#676A86"),
)

styles.save(chart, path=".", name="public_sector_share", svg=True)
chart.save("public_sector_share.png", scale_factor=2)
chart
```
