```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles()
styles.register_and_enable_theme()

# Source: CSconnected SIPF Impact Report 2026 (WERU, Cardiff University).
# total Welsh jobs = direct + modelled indirect.
# 2020 = baseline · 2025 = latest actual · 2030 = CSconnected ambition.
df = pd.DataFrame({
    "year":  ["2020", "2025", "2030"],
    "value": [2085, 3140, 6000],
    "kind":  ["Actual", "Actual", "Target"],
})
df["label"] = df["value"].map(lambda v: f"{v:,.0f}")

base = alt.Chart(df).encode(
    x=alt.X("year:N", title=None, scale=alt.Scale(domain=["2020", "2025", "2030"], paddingInner=0.45)),
    y=alt.Y("value:Q", title="Total jobs supported in Wales"),
)

bars = base.mark_bar().encode(
    color=alt.Color("kind:N", scale=alt.Scale(domain=["Actual", "Target"]), legend=None))
labels = base.mark_text(dy=-8, fontWeight="bold").encode(text="label:N")

caption = alt.Title(
    text="Source: CSconnected SIPF Impact Report 2026 (WERU, Cardiff University)",
    subtitle=[
        "2020 and 2025 measured; 2030 is the cluster's stated ambition.",
    ],
    orient="bottom", anchor="start", dy=12,
)

chart = (bars + labels).properties(
    width=560, height=340, title=caption,)

styles.save(chart, path=".", name="cs_jobs_2030", svg=True)
chart.save("cs_jobs_2030.png", scale_factor=2)
chart
```
