```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles()
styles.register_and_enable_theme()

# Start-ups per 1M population, top 20 European countries.
# Source: Atomico, State of European Tech (data: Dealroom.co). Excludes Israel;
# countries with population over 1M. European average = 190.
data = [("Estonia", 865), ("Ireland", 666), ("Denmark", 573), ("Finland", 525),
        ("Netherlands", 507), ("Sweden", 429), ("Switzerland", 420),
        ("United Kingdom", 406), ("Lithuania", 390), ("Germany", 366),
        ("Norway", 283), ("France", 253), ("Cyprus", 246), ("Italy", 234),
        ("Belgium", 211), ("Latvia", 190), ("Slovenia", 166), ("Spain", 157),
        ("Austria", 145), ("Portugal", 130)]
df = pd.DataFrame(data, columns=["country", "val"])
order = [c for c, _ in data]
df["hl"] = df["country"].eq("United Kingdom")

base = alt.Chart(df).encode(
    x=alt.X("country:N", title=None, sort=order, axis=alt.Axis(labelAngle=-45)),
    y=alt.Y("val:Q", title="Start-ups per 1M population"),
)

# colours come from the EcoStyles palette; domain order fixes which bar gets which.
# [False, True] -> non-highlight bars take the theme's 1st colour (teal),
# UK (hl=True) takes the 2nd (red). Flip the domain if they come out reversed.
bars = base.mark_bar().encode(
    color=alt.Color("hl:N", scale=alt.Scale(domain=[False, True]), legend=None))

avg = alt.Chart(pd.DataFrame({"y": [190]})).mark_rule(
    strokeDash=[4, 3]).encode(y="y:Q")

avg_label = alt.Chart(pd.DataFrame({"y": [190], "t": ["European average"]})).mark_text(
    align="right", dx=-4, dy=-6).encode(
    y="y:Q", x=alt.value(640), text="t")

chart = (bars + avg + avg_label).properties(
    width=640, height=330,
    title=alt.Title(
        text="Start-ups per million population, top European countries",
        anchor="start",
        subtitle="Source: Atomico, State of European Tech (Dealroom.co)"),
)

styles.save(chart, path=".", name="startup_density", svg=True)
chart.save("startup_density.png", scale_factor=2)
chart
```
