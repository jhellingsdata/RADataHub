```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles()
styles.register_and_enable_theme()

RED, GREY = "#E6224B", "#94a3b8"

# Computed from HESA sb275 Figure 15 (2023/24): share of each nation's
# domiciled graduates working outside their country of permanent address.
df = pd.DataFrame({
    "nation": ["Wales", "Northern Ireland", "Scotland", "England"],
    "pct":    [27, 15, 10, 2],
})
df["hl"] = df["nation"].eq("Wales")
df["label"] = df["pct"].map(lambda v: f"{v}%")

# explicit top-to-bottom order: England (smallest) at top, Wales (largest) at bottom
order = ["England", "Scotland", "Northern Ireland", "Wales"]

base = alt.Chart(df).encode(
    y=alt.Y("nation:N", title=None, scale=alt.Scale(domain=order)),
    x=alt.X("pct:Q", title="% working outside home nation", scale=alt.Scale(domain=[0, 30])),
)

bars = base.mark_bar().encode(
    color=alt.Color("hl:N", scale=alt.Scale(domain=[True, False], range=[RED, GREY]), legend=None))

labels = base.mark_text(dx=5, align="left", fontWeight="bold").encode(text="label:N")

caption = alt.Title(
    text="Source: HESA Graduate Outcomes 2023/24 (sb275, Figure 15)",
    subtitle=[
        "Share of graduates working outside their country of permanent address.",
        "More than a quarter of Welsh graduates leave — far above other UK nations.",
    ],
    orient="bottom", anchor="start", dy=12,
)

chart = (bars + labels).properties(
    width=560, height=240, title=caption,
    padding={"bottom": 40, "left": 5, "right": 5, "top": 5},
)

styles.save(chart, path=".", name="wales_graduate_outflow", svg=True)
chart.save("wales_graduate_outflow.png", scale_factor=2)
chart
```
