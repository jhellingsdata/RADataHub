```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles

styles = EcoStyles()
styles.register_and_enable_theme()

# Estimated additional data-centre-driven wholesale electricity cost per average
# Irish household, € per year, 2015–2023 (cumulative €360).
years = list(range(2015, 2024))
cost = [23.63, 6.67, 7.72, 14.00, 21.54, 23.26, 66.82, 130.16, 66.44]
df = pd.DataFrame({"year": years, "eur": cost})
df["lab"] = df["eur"].map(lambda v: f"€{v:.0f}")

bars = alt.Chart(df.assign(series="Cost")).mark_bar().encode(
    x=alt.X("year:O", title=None),
    y=alt.Y("eur:Q", title="Additional cost per household, €"),
    color=alt.Color("series:N", legend=None),   # theme palette assigns its 1st colour
)

labels = alt.Chart(df).mark_text(dy=-6, fontWeight="bold").encode(
    x="year:O", y="eur:Q", text="lab")

chart = (bars + labels).properties(
    width=560, height=300,
    padding={"bottom": 20, "left": 5, "right": 5, "top": 5},
    title=alt.Title(
        text="Estimated data centre cost per Irish household, € per year, 2015–2023",
        anchor="start",
        subtitle="Source: Friends of the Earth (2025)"),
)

styles.save(chart, path=".", name="ireland_household_cost", svg=True)
chart.save("ireland_household_cost.png", scale_factor=2)
chart
```
