```python
import pandas as pd
import altair as alt
from ecostyles import EcoStyles   # your package
```

```python

# ---- data ----
data = [
 ("Japan",204.4),("Singapore",171.9),("Sudan",169.1),("Bahrain",152.4),
 ("Italy",138.4),("Greece",136.9),("Senegal",132.3),("Maldives",129.4),
 ("United States",125.8),("Ukraine",122.6),("Bhutan",120.3),
 ("St Vincent & Gren.",120.1),("France",118.4),("Canada",110.7),
 ("Belgium",109.2),("China",106.9),("Mozambique",106.1),
 ("United Kingdom",103.6),("Bolivia",102.7),
]
df = pd.DataFrame(data, columns=["country","debt"])

# ---- enable EO theme ----
styles = EcoStyles()
styles.register_and_enable_theme()      # match however you call it in your other charts

# ---- chart ----
bars = (
    alt.Chart(df).mark_bar()
    .encode(
        y=alt.Y("country:N", sort="-x", title=None,
                axis=alt.Axis(labelLimit=200, labelFontSize=10)),
        x=alt.X("debt:Q", title="% of GDP", scale=alt.Scale(domain=[0,220]),
                axis=alt.Axis(grid=False)),
        tooltip=["country","debt"],
    )
)
rule = (alt.Chart(pd.DataFrame({"x":[100]})).mark_rule(strokeDash=[4,3]).encode(x="x:Q"))

chart = (bars + rule).properties(
    width=600, height=460,
    title=alt.Title("Nineteen economies now carry public debt above 100% of GDP",
                    subtitle="General government gross debt (% of GDP), 2026 projections"),
)

styles.save(chart, name="debt_over_100", source="IMF World Economic Outlook", svg=True)
chart
```
